# TTS Voice Wizard

## Project Overview
This project demonstrates a Text-to-Speech (TTS) application inspired by the TTS Voice Wizard project. The system converts written text into spoken audio using Python-based Text-to-Speech technology.

- Allows users to enter custom text

```python
import pyttsx3

engine = pyttsx3.init()

# Get available voices
voices = engine.getProperty('voices')

print("Available voices:")
for i, voice in enumerate(voices):
    print(f"{i}: {voice.name}")

# Set speech rate
engine.setProperty('rate', 150)

# Set volume (0.0 to 1.0)
engine.setProperty('volume', 1.0)

print("\nVoice settings configured successfully!")
```

```python
import pyttsx3

# Initialize the TTS engine
engine = pyttsx3.init()

# Get text from the user
text = input("Enter the text you want to convert to speech: ")

# Configure voice settings
engine.setProperty("rate", 150)
engine.setProperty("volume", 1.0)

# Convert text to speech
engine.say(text)
engine.runAndWait()

print("Your custom text was converted to speech successfully!")
```

```python
import pyttsx3

engine = pyttsx3.init()
voices = engine.getProperty("voices")

print("Available voices:")
for i, voice in enumerate(voices):
    print(f"{i}: {voice.name}")

voice_choice = int(input("\nEnter the voice number you want to use: "))
text = input("Enter the text you want to convert to speech: ")

# Set selected voice
engine.setProperty("voice", voices[voice_choice].id)

# Voice settings
engine.setProperty("rate", 150)
engine.setProperty("volume", 1.0)

# Speak the text
engine.say(text)
engine.runAndWait()

print("\nSpeech generated successfully with the selected voice!")
```

```python
import pyttsx3

# Initialize the Text-to-Speech engine
engine = pyttsx3.init()

# Get user input
text = input("Enter the text you want to save as speech: ")

# Set voice properties
engine.setProperty("rate", 150)
engine.setProperty("volume", 1.0)

# Save speech as an audio file
engine.save_to_file(text, "tts_output.mp3")
engine.runAndWait()

print("Audio file saved successfully as tts_output.mp3")
```
